## Part 2: Data Preparation & Train-Test Split

**The Problem:** Before training any models, we need to:
1. Handle missing values
2. Encode categorical variables
3. Scale numeric features (important for Logistic Regression)
4. Split our data into training and test sets

### Task 2a: Prepare Features

Create a function that:
- Drops `customer_id` (not predictive)
- Fills missing values in `age`, `website_visits_monthly`, and `email_click_rate` with their median
- Encodes `is_member` (Yes=1, No=0)
- Encodes `customer_tier` ordinally (Bronze=0, Silver=1, Gold=2, Platinum=3)
- One-hot encodes `region`
- Scales numeric features using StandardScaler

````python
def prepare_features(df):
    """
    Prepare features for machine learning.
    
    Steps:
    1. Drop customer_id
    2. Fill missing values with median
    3. Encode is_member (Yes=1, No=0)
    4. Encode customer_tier ordinally (Bronze=0, Silver=1, Gold=2, Platinum=3)
    5. One-hot encode region
    6. Scale numeric features using StandardScaler
    
    Args:
        df (pd.DataFrame): The raw DataFrame
        
    Returns:
        pd.DataFrame: The prepared DataFrame with all numeric columns (scaled)
    """
    # IMPORTANT: Make a copy first so we don't modify the original!
    df_prep = df.copy()
    
    # Your code here
    # Step 1: Drop customer_id
    df_prep = df_prep.drop(columns=['customer_id'], errors='ignore')

    # Step 2: Fill missing values with median (only for numeric columns)
    df_prep = df_prep.fillna(df_prep.median(numeric_only=True))

    # Step 3: Encode is_member (Yes=1, No=0)
    df_prep['is_member'] = df_prep['is_member'].map({'Yes': 1, 'No': 0})

    # Step 4: Encode customer_tier ordinally (Bronze=0, Silver=1, Gold=2, Platinum=3)
    tier_mapping = {'Bronze': 0, 'Silver': 1, 'Gold': 2, 'Platinum': 3}
    df_prep['customer_tier'] = df_prep['customer_tier'].map(tier_mapping)

    # Step 5: One-hot encode region
    df_prep = pd.get_dummies(df_prep, columns=['region'], prefix='region')
    
    # Step 6: Scale numeric features using StandardScaler
    # Scale only continuous numeric features; exclude ordinal, binary, and target columns
    scaler = StandardScaler()
    exclude_cols = {'customer_tier', 'is_member', 'annual_spend', 'churned'}
    numeric_cols = [col for col in df_prep.select_dtypes(include=['int64', 'float64']).columns
                    if col not in exclude_cols]
    df_prep[numeric_cols] = scaler.fit_transform(df_prep[numeric_cols])
        
    return df_prep
````

### Task 2b: Split Data for Training and Testing

Create a function that splits the data into training and test sets.

**Important:** When splitting for classification (predicting `churned`), use `stratify=y` to maintain the same class balance in both sets.

````python
def split_data(df, target_column, test_size=0.2, random_state=42):
    """
    Split data into training and test sets.
    
    Args:
        df (pd.DataFrame): The prepared DataFrame
        target_column (str): Name of the target column ('annual_spend' or 'churned')
        test_size (float): Proportion of data for testing (default 0.2)
        random_state (int): Random seed for reproducibility (default 42)
        
    Returns:
        tuple: (X_train, X_test, y_train, y_test)
    
    Note: Drop BOTH target columns from X (annual_spend AND churned).
    For classification targets, use stratify to maintain class balance.
    """
    # Your code here
    X = df.drop(columns=['annual_spend', 'churned'])
    y = df[target_column]
    
    stratify = y if target_column == 'churned' else None
    
    return train_test_split(X, y, test_size=test_size, random_state=random_state, stratify=stratify)
````

> **Please answer the question below. It is a required part of the assignment!**

**Question:** Why do we split the data into training and test sets BEFORE training any models? What could go wrong if we trained on all the data and then tested on the same data?

**Answer:** 

We split the data before training to simulate how the model will perform on *unseen, real-world data*. 

The test set acts as an honest, unbiased evaluation — the model has never seen it during training, so its performance on that set reflects how well it has truly learned the underlying patterns.

If we trained on all the data and then tested on the same data, we would be measuring how well the model *memorized* the training data rather than how well it *generalizes*. This is called **data leakage** and leads to **overfitting** — the model appears to perform extremely well on paper, but fails badly when deployed on new customers. 

Essentially, we would be "teaching to the test" and getting a falsely optimistic picture of the model's real usefulness.

## Part 4: Regression Evaluation

**The Problem:** How do we know if our predictions are good? We need metrics!

### Key Regression Metrics:
- **MSE (Mean Squared Error)**: Average of squared errors. Penalizes large errors heavily.
- **RMSE (Root Mean Squared Error)**: Square root of MSE. Same units as target variable.
- **MAE (Mean Absolute Error)**: Average of absolute errors. More robust to outliers.
- **R² (R-Squared)**: Proportion of variance explained. 1.0 is perfect, 0 is no better than guessing the mean.

### Task: Create an Evaluation Function

````python
def evaluate_regression(y_true, y_pred):
    """
    Evaluate regression predictions using multiple metrics.
    
    Args:
        y_true: Actual values
        y_pred: Predicted values
        
    Returns:
        dict: Dictionary with keys 'mse', 'rmse', 'mae', 'r2'
    """
    # Your code here
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    
    return {
        'mse': mse,
        'rmse': rmse,
        'mae': mae,
        'r2': r2
    }    
````

**Question:** How is logistic regression different from linear regression, even though both have "regression" in the name?

**Answer:** 

Despite sharing the word "regression," they solve fundamentally different types of problems:

- **Linear regression** predicts a *continuous numerical value* (e.g., how much a customer will spend). Its output can be any number along a continuous scale.
- **Logistic regression** predicts the *probability of belonging to a category* (e.g., will a customer churn: Yes or No). Its output is always between 0 and 1, thanks to the **sigmoid (logistic) function** that maps any input to that range. We then apply a threshold (usually 0.5) to assign a class label.

In short: linear regression answers "how much?", while logistic regression answers "which category?" or "what is the probability of X?". The "regression" in logistic regression refers to the underlying mathematical technique of fitting a line, not the type of output.

## Part 6: Classification Evaluation

**The Problem:** For classification, we need different metrics than regression.

### Key Classification Metrics:
- **Accuracy**: % of correct predictions. Can be misleading with imbalanced classes!
- **Precision**: Of all predicted churners, how many actually churned? (Avoid false alarms)
- **Recall**: Of all actual churners, how many did we catch? (Don't miss real churners)
- **F1 Score**: Harmonic mean of precision and recall. Good balance.

### Task: Create Classification Evaluation Functions

````python
def evaluate_classification(y_true, y_pred):
    """
    Evaluate classification predictions using multiple metrics.
    
    Args:
        y_true: Actual labels (0 or 1)
        y_pred: Predicted labels (0 or 1)
        
    Returns:
        dict: Dictionary with keys 'accuracy', 'precision', 'recall', 'f1'
    """
    # Your code here
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred)
    recall = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)
    
    return {
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1': f1
    }
````

````python
def plot_confusion_matrix(y_true, y_pred, title='Confusion Matrix'):
    """
    Plot a confusion matrix heatmap.
    
    Args:
        y_true: Actual labels
        y_pred: Predicted labels
        title: Plot title
    """
    # Your code here
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=['Predicted 0', 'Predicted 1'], 
                yticklabels=['Actual 0', 'Actual 1'],
                cbar=False)
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.title(title)
    plt.tight_layout()
    plt.show()
````

**Answer 1:** 

**Recall is more important** for ShopSmart in this context. Here is why:
- A **false negative** (predicting a customer will *stay* when they actually *churn*) means ShopSmart does nothing and loses the customer entirely — that revenue is gone with no chance to intervene.
- A **false positive** (predicting a customer will *churn* when they actually *stay*) means ShopSmart offers a retention incentive (e.g., a discount) to a customer who was not going to leave anyway. This wastes a small amount of money, but is a much smaller cost than losing the customer permanently.

Since the cost of missing a churner is much higher than the cost of offering an unnecessary discount, we want to cast a wide net and catch as many at-risk customers as possible — which is exactly what high recall does.

**Question 2:** If we predicted "no churn" for everyone, what would our accuracy be? Why is this misleading?

**Answer 2:**

But this is completely misleading because the model is utterly useless: it never identifies a single at-risk customer, which defeats the entire purpose of the project. This is why accuracy alone is a poor metric for imbalanced datasets. Metrics like **F1-score, precision, and recall** give a more honest picture of model performance when one class is much rarer than the other.

**Question 1:** Why does Random Forest often outperform a single decision tree?

**Answer 1:** 

A single decision tree is prone to **overfitting** — it tends to learn the training data too precisely, including its noise, which hurts performance on new data. Random Forest addresses this through two key mechanisms:
1. **Bagging (Bootstrap Aggregating):** It trains many trees, each on a different random subset of the training data (sampled with replacement). Each tree is slightly different, and averaging their predictions reduces variance.
2. **Feature randomness:** At each split, only a random subset of features is considered. This ensures the trees are decorrelated from each other — they make different mistakes, so when combined, the errors cancel out.

The result is an ensemble that is much more robust and generalizes far better than any single tree.

### Key Hyperparameters:
- **n_estimators**: Number of boosting stages (trees built sequentially)
- **learning_rate**: How much each tree contributes (smaller = more trees needed but often better)

### Task: Train and Tune Gradient Boosting

````python
def train_gradient_boosting(X_train, y_train, n_estimators=100, learning_rate=0.1):
    """
    Train a gradient boosting classifier.
    
    Args:
        X_train (pd.DataFrame): Training features
        y_train (pd.Series): Training target
        n_estimators (int): Number of boosting stages (default 100)
        learning_rate (float): Learning rate (default 0.1)
        
    Returns:
        GradientBoostingClassifier: The trained model
    """
    # Your code here
    model = GradientBoostingClassifier(n_estimators=n_estimators, learning_rate=learning_rate, random_state=42)
    model.fit(X_train, y_train)
    
    return model
````

### Written Reflection

**Question 1:** What's the key difference between **bagging** (Random Forest) and **boosting** (Gradient Boosting)?

**Answer 1:** 

The core difference is in *how* the trees are combined:

- **Bagging (Random Forest):** Trees are built *independently and in parallel*, each on a random sample of the data. Their predictions are then averaged (or majority-voted). The goal is to reduce variance by combining many diverse, independently-built models.
- **Boosting (Gradient Boosting):** Trees are built *sequentially*, where each new tree focuses on correcting the *errors made by all previous trees*. The model learns iteratively — weak learners are combined in a way that each one nudges the overall prediction closer to the truth. The goal is to reduce bias by progressively improving on mistakes.

In short: bagging builds trees in parallel to reduce variance; boosting builds trees in sequence to reduce bias.

**Question 2:** Which ensemble method performed better on this dataset? Why might that be?

**Answer 2:** 

Gradient Boosting typically achieves a higher F1-score on this dataset because it is specifically designed to minimize the loss function step-by-step, allowing it to model complex, subtle patterns in customer behavior more precisely. Since churn prediction involves nuanced interactions between features (e.g., a drop in purchase frequency combined with low recent engagement), boosting's sequential error-correction is well-suited for the task. 

Random Forest, while more robust and faster to train, is a blunter instrument — it averages independent trees rather than strategically targeting remaining errors.
